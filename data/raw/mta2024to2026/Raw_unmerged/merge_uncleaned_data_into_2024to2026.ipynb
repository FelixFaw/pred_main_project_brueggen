```python
import pandas as pd
from pathlib import Path

file_list = [
    r'C:\Users\golde\PycharmProjects\pred_main_project_brueggen\data\raw\mta2024to2026\Raw_unmerged\STW-Mittelteilanlage 2024.xlsx', 
    r'C:\Users\golde\PycharmProjects\pred_main_project_brueggen\data\raw\mta2024to2026\Raw_unmerged\Störliste STW-Mittelteilanlage 2025.xlsx',
    r'C:\Users\golde\PycharmProjects\pred_main_project_brueggen\data\raw\Störliste STW-Mittelteilanlage 2026.xlsx'
]

df_list = []

for file in file_list:
    path_obj = Path(file)
    try:
        # header=0 bedeutet, die erste Zeile enthält die Spaltennamen
        df = pd.read_excel(path_obj, header=0)
        
        # WICHTIG: Entferne Spalten, die nur "Unnamed" heißen (entstehen oft durch leere Zellen)
        df = df.loc[:, ~df.columns.str.contains('^Unnamed')]
        
        # Spaltennamen bereinigen (Zeilenumbrüche entfernen)
        df.columns = [str(col).replace('\n', ' ').strip() for col in df.columns]
        
        df_list.append(df)
        print(f"Erfolgreich geladen: {path_obj.name}")
    except Exception as e:
        print(f"Fehler bei {file}: {e}")

if df_list:
    # Zusammenfügen
    result_df = pd.concat(df_list, axis=0, ignore_index=True)
    
    # Speichern als CSV
    output_path = r'C:\Users\golde\PycharmProjects\pred_main_project_brueggen\data\zusammengefuehrtes_file.csv'
    result_df.to_csv(output_path, index=False, sep=';', encoding='utf-8-sig')
    
    print(f"\nErfolg! {len(result_df)} Zeilen in '{output_path}' gespeichert.")
```
