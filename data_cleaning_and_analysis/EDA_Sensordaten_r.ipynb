```python
import os

import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Basis-Pfad
base_path = r"C:\Users\buntjerr\IdeaProjects\untitled\pred_main_project_brueggen\data\unprocessed_r"

# Dateien laden
dferrors = pd.read_csv(
    os.path.join(base_path, "AHF_errors.csv"),
    sep=",",
    dtype="string",
    na_values=["", " ", "  "],
    low_memory=False
)

dfmachine = pd.read_csv(
    os.path.join(base_path, "AHF_machineState.csv"),
    sep=";",
    dtype="string",
    na_values=["", " ", "  "],
    low_memory=False
)

dforder = pd.read_csv(
    os.path.join(base_path, "AHF_orderState.csv"),
    sep=";",
    dtype="string",
    na_values=["", " ", "  "],
    low_memory=False
)

dfresources = pd.read_csv(
    os.path.join(base_path, "AHF_resources.csv"),
    sep=",",
    dtype="string",
    na_values=["", " ", "  "],
    low_memory=False
)

# Kontrolle
for i, df in enumerate([dferrors, dfmachine, dforder, dfresources], start=1):
    print(f"DF{i} Shape:", df.shape)
```

```python
dfmachine["__time"] = pd.to_datetime(
    dfmachine["__time"],
    dayfirst=True,
    errors="coerce"
)
dforder["__time"] = pd.to_datetime(
    dforder["__time"],
    dayfirst=True,
    errors="coerce"
)
```

```python
state_time.columns = [str(col)[:10] for col in state_time.columns]
```

```python
state_time.plot(
    kind="area",
    stacked=True,
    figsize=(12, 6),
    colormap="tab20"
)
plt.xlabel("Zeit")
plt.ylabel("Anzahl Events")
plt.title("Top 5 Zustände über Zeit (stündlich)")
plt.legend(title="State", bbox_to_anchor=(1.05, 1), loc="upper left")

plt.tight_layout()
plt.show()
```

```python
state_time_norm = state_time.div(state_time.sum(axis=1), axis=0)

state_time_norm.plot(    kind="area",
                         stacked=True,
                         figsize=(12, 6),
                         colormap="tab20"
                         )

plt.xlabel("Zeit")
plt.ylabel("Anteil")
plt.title("Relative Verteilung der Zustände über die Zeit")
plt.legend(title="State", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.show()
```
